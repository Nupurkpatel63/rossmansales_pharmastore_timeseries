```python
%%writefile app.py

import streamlit as st
import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt
import os

# -----------------------------
# Page configuration
# -----------------------------

st.set_page_config(
    page_title="Store Sales & Customer Prediction",
    page_icon="📊",
    layout="wide"
)

st.title("📊 Store Sales & Customer Prediction")
st.write(
    "Enter store information or upload a CSV file to generate predictions."
)

# -----------------------------
# Load models
# -----------------------------

MODEL_PATH = "/content/drive/MyDrive/Colab Notebooks/project6/deployment/models"

sales_model = joblib.load(
    os.path.join(MODEL_PATH, "sales_model_gb.pkl")
)

customer_model = joblib.load(
    os.path.join(MODEL_PATH, "customer_model_gb.pkl")
)

st.success("Sales and Customer models loaded successfully.")

# -----------------------------
# Store ID
# -----------------------------

store_id = st.number_input(
    "Store ID",
    min_value=1,
    value=1,
    step=1
)

# -----------------------------
# CSV Upload
# -----------------------------

uploaded_file = st.file_uploader(
    "Upload prediction CSV",
    type=["csv"]
)

if uploaded_file is not None:

    df = pd.read_csv(uploaded_file)

    st.subheader("Uploaded Data")

    st.dataframe(df)

    st.write("Columns detected:")
    st.write(df.columns.tolist())

    # -------------------------
    # Add Store ID
    # -------------------------

    df["Store_id"] = store_id

    # -------------------------
    # Prediction button
    # -------------------------

    if st.button("Generate Predictions"):

        try:

            sales_pred = sales_model.predict(df)

            customer_pred = customer_model.predict(df)

            result = df.copy()

            result["Predicted_Sales"] = sales_pred
            result["Predicted_Customers"] = customer_pred

            st.subheader("Prediction Results")

            st.dataframe(result)

            # -----------------
            # Plot Sales
            # -----------------

            st.subheader("Predicted Sales")

            fig1, ax1 = plt.subplots()

            ax1.plot(
                result["Predicted_Sales"],
                marker="o"
            )

            ax1.set_xlabel("Prediction")
            ax1.set_ylabel("Sales")
            ax1.set_title("Predicted Sales")

            st.pyplot(fig1)

            # -----------------
            # Plot Customers
            # -----------------

            st.subheader("Predicted Customers")

            fig2, ax2 = plt.subplots()

            ax2.plot(
                result["Predicted_Customers"],
                marker="o"
            )

            ax2.set_xlabel("Prediction")
            ax2.set_ylabel("Customers")
            ax2.set_title("Predicted Customer Numbers")

            st.pyplot(fig2)

            # -----------------
            # Download CSV
            # -----------------

            csv = result.to_csv(index=False)

            st.download_button(
                label="Download Predictions CSV",
                data=csv,
                file_name="store_predictions.csv",
                mime="text/csv"
            )

        except Exception as e:

            st.error(
                f"Prediction failed: {str(e)}"
            )
```
